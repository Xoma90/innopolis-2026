```python
import random
import statistics

#  Функция генерации списка случайных чисел в диапазоне от a до b длинной length
def generate_random_list(a: int, b: int, length: int) -> list:
    return [random.randint(a, b) for _ in range(length)]

# Функция, строящая частотный ряд в виде словаря
def build_frequency_dict(lst: list) -> dict:
    frequency = {}
    for item in lst:
        frequency[item] = frequency.get(item, 0) + 1
    return frequency

# Функция счета минимума, среднего, медианы, моды и максимума для значений списка
def calculate_statistics(lst: list) -> dict:
    if not lst:
        return {}
        
    return {
        "Минимум (min)": min(lst),
        "Среднее (mean)": statistics.mean(lst),
        "Медиана (median)": statistics.median(lst),
        "Мода (mode)": statistics.mode(lst),
        "Максимум (max)": max(lst)
    }

if __name__ == "__main__":

# Выводим результаты
    my_list = generate_random_list(a=1, b=10, length=10)
    print(f"1. Созданный список: {my_list}")
    
    freq_dict = build_frequency_dict(my_list)
    print(f"2. Частотный ряд: {freq_dict}")
    
    stats = calculate_statistics(my_list)
    print(f"3. Статистика:")
    for metric, value in stats.items():
        print(f"   - {metric}: {value}")
```
